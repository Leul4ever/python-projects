```python
# SQLite is a lightweight, disk-based database that doesn't require a separate server process.
import sqlite3 as db
import pandas as pd
```

```python
con = db.connect('sqlite.db')
## cursor userd to execute SQL commands
cn = con.cursor()
```

```python
# drop table if exists
cn.execute("DROP TABLE IF EXISTS employees")
```

```python
cn.execute("CREATE TABLE employees (empid INTEGER PRIMARY KEY, firstname NVARCHAR(50), lastname NVARCHAR(50))")
```

```python
cn.execute("SELECT name FROM sqlite_master WHERE type='table';")
print(cn.fetchall())
```

```python
cn.execute("INSERT OR IGNORE INTO employees VALUES (1, 'John', 'Doe')")
cn.execute("INSERT OR IGNORE INTO employees VALUES (2, 'Jason', 'Hale')")
cn.execute("INSERT OR IGNORE INTO employees VALUES (3, 'Sam', 'Billing')")


```

```python

cn.execute("SELECT * FROM employees;")
print(cn.fetchall())
```
