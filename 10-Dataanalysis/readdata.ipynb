Readin Data from Different Source 


```python
from io import StringIO
import pandas as pd

data = """
{
    "employee_name": "leul abera",
    "email": "leulabera70@gmail.com",
    "job_profile": [
        {
            "title1": "Team Lead",
            "title2": "Sr Developer"
        }
    ]
}
"""

# Convert string to file-like
buffer = StringIO(data)

# Read JSON
df = pd.read_json(buffer)

print(df)
```
