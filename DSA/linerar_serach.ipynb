```python
def linerarSerach(list,target):
    """
    return the index position of the target if found , else return none
    """
    for i in range(0,len(list)):
        if list[i]==target:
            return i 
    return None
def verify(index):
    if index is not None:
        print("Target found at index:",index)
    else:
        print("Target not found in list")
numbers=[1,2,3,4,5,6]
result =linerarSerach(numbers, 6)
verify(result)
result =linerarSerach(numbers, 12)
verify(result)

```
